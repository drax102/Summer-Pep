```python
inventory = ["Mouse", "Keyboard", "Monitor", "CPU", "Speaker"]

inventory.sort()
print("Alphabetical:", inventory)

inventory.reverse()
print("Reverse Alphabetical:", inventory)
```

```python
original = ["File1", "File2", "File3"]

backup = original.copy()

backup.append("File4")

print("Original:", original)
print("Backup:", backup)
```

```python
employees = ["Aman", "Rahul", "Neha", "Rahul", "Priya"]

print("Aman Position:", employees.index("Aman"))
print("Rahul Occurrences:", employees.count("Rahul"))
```

```python
alerts = ["Fire", "Power Failure", "Network Down"]

print("Before:", alerts)

alerts.clear()

print("After: No alerts found", alerts)
```
