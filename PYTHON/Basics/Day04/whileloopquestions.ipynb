```python
#no of digits count
n=int(input("Enter your number:"))
count=0
while n>0:
    count+=1
    n=n//10

print("No of Digits:",count)
```

```python
#reverse a number
n=int(input("Enter your number:"))
reverse=0
while n>0:
    digit=n%10
    reverse=reverse*10+digit
    n=n//10
print("reverse is: ", reverse)
```

```python
#palindrome number
n=int(input("Enter a number: "))
a=n
reverse=0
while n>0:
    digit=n%10
    reverse=reverse*10+digit
    n=n//10
if(a==reverse):
    print("palindrome")
else:
    print("No Palindrome")
```

```python
# continue the take input till the user inputs a zero
n=int(input("Enter your number:"))
while n!=0:
    n=int(input("Enter your number:"))
if n==0:
    print("you have enetered zero")
```

```python
# to print row1
#column1 column2 column3
#row 2
#column1 column2 column3
#row 3
#column1 column2 column3

for i in range(1,4):
    print("row",i)
    for j in range(1,i+1):
        print("column",j)
```
