```python
Marks = {
    "Harsh": 99,"Shiva": 49,"Rajat": 69,"Aman": 85,"Priya": 78
}

average = sum(Marks.values()) / len(Marks)

print("Class Average:", average)

print("Students scoring above average:")

for name, mark in Marks.items():
    if mark > average:
        print(name, ":", mark)
```

```python
#1)Design a mini social-network data model supporting friends and mutual friends and friend suggestions0
# 2) Build a recommendation engine that suggests products products purchased by similar customers but not by the current customer.
# 3)create an airline reservation model and detect double-booked seats.
# 4)Develop an online coding contest scoreboard supporting rank updates after every submission.
# Build a Plagiarism checker that compares two assignments based on unique words and similarity percentage.
# Mini Social Network

friends = {
    "Rahul": {"Aman", "Priya", "Riya", "Karan"},
    "Aman": {"Rahul", "Riya", "Karan", "Neha"},
    "Priya": {"Rahul", "Riya", "Sneha"},
    "Riya": {"Rahul", "Aman", "Priya", "Neha"},
    "Karan": {"Rahul", "Aman", "Rohit"},
    "Neha": {"Aman", "Riya"},
    "Sneha": {"Priya"},
    "Rohit": {"Karan"}
}

print("Rahul's Friends:", friends["Rahul"])

print("Aman's Friends:", friends["Aman"])

print("\nMutual Friends of Rahul and Aman:")
print(friends["Rahul"] & friends["Aman"])

print("\nFriend Suggestions for Rahul:")
print(friends["Aman"] - friends["Rahul"] - friends["Neha"])
```
