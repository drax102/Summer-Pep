```python
# Netflix content dashbord indexing & slicing . display first, last, top 3, last 3 , alternate shows, reverse using slicing.

netflix_shows=['obsession','50 shades of grey','365 days','the girl next door','cruel intention','darker grey']
print("Original List:", netflix_shows)
print("-" * 40)
print("First show:", netflix_shows[0])
print("Last show:", netflix_shows[-1])
print("Top 3 shows:", netflix_shows[:3])
print("Last 3 shows:", netflix_shows[-3:])
print("Alternate shows:", netflix_shows[::2])
print("Reversed shows:", netflix_shows[::-1])
```
