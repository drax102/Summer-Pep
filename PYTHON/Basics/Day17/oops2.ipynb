```python
# Create a square and triangle classes, each other with a draw () method that prints a different simple pattern (eg using *) Put objects of both in a list loop through calling draw() on each.


class Square:
    def draw(self):
        print("Square")
        print("****")
        print("*  *")
        print("*  *")
        print("****")


class Triangle:
    def draw(self):
        print("Triangle")
        print("  *")
        print(" ***")
        print("*****")



shapes = [Square(), Triangle()]

for shape in shapes:
    shape.draw()
    print()
```
