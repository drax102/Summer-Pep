```python
class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price

    def display(self):
        print("Title :", self.title)
        print("Author:", self.author)
        print("Price :", self.price)


class Library:
    def __init__(self):
        self.books = []

    def add_book(self, book):
        self.books.append(book)

    def show_all_books(self):
        print("Books in Library:")
        for book in self.books:
            book.display()
            print()


b1 = Book("Python Basics", "Guido", 500)
b2 = Book("Java Programming", "James Gosling", 700)
b3 = Book("C Programming", "Dennis Ritchie", 600)

library = Library()

library.add_book(b1)
library.add_book(b2)
library.add_book(b3)

library.show_all_books()

b4 = Book("Absuidj", "sihfij", 199)
b4.display()
```

```python
class Student:
    def __init__(self,roll_no,name,marks):
        self.roll_no=roll_no
        self.name=name
        self.marks=marks
    def markscheck(self):
        if(self.marks>90):
            print(self.name)
            print("A") 
            print(self.roll_no)
        elif(self.marks<90 and self.marks>75):
            print("b")
            print(self.name)
            print(self.roll_no)
        else:
            print("C")
s1=Student(55,"utkarsh",88)
s1.markscheck()
```

```python
class Rectangle:
    def __init__(self, length,breadth):
        self.length=length
        self.breadth=breadth
    def area(self):
        return self.length*self.breadth
    def perimeter(self):
        return 2*(self.length+self.breadth)
s1=Rectangle(6,98)
print(s1.area())
print(s1.perimeter())
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def give_raise(self, percent):
        self.salary = self.salary + (self.salary * percent / 100)
        print("Updated Salary:", self.salary)

e1 = Employee("Utkarsh", 50000)
print("Name:", e1.name)
print("Salary:", e1.salary)
e1.give_raise(10)
```
