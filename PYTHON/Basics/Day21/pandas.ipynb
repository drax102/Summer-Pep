```python
import pandas as pd

s = pd.Series([10, 20, 30, 40, 50])

print(s.head(3))      # First 3 values
print(s.tail(2))      # Last 2 values
print(s.size)         # Number of elements
print(s.count())      # Count non-null values
print(s.sum())        # Sum
print(s.mean())       # Average
print(s.max())        # Maximum
print(s.min())        # Minimum
print(s.median())     # Median
print(s.std())        # Standard deviation
print(s.describe())   # Statistical summary
```

| **Error**                         | **Cause**                                        | **Solution**                                                          |
| --------------------------------- | ------------------------------------------------ | --------------------------------------------------------------------- |
| `FileNotFoundError`               | File path or file name is incorrect              | Check the file name and path. Use the correct location.               |
| `PermissionError`                 | File is open in another program or no permission | Close the file (Excel/CSV) or run with proper permissions.            |
| `ModuleNotFoundError: openpyxl`   | `openpyxl` package is not installed              | Install using `pip install openpyxl`.                                 |
| `UnicodeDecodeError`              | Wrong file encoding                              | Specify encoding, e.g., `pd.read_csv("file.csv", encoding="latin1")`. |
| `ParserError`                     | Incorrect delimiter or corrupted CSV             | Use the correct separator, e.g., `sep=","` or `sep=";"`.              |
| `ValueError: Worksheet not found` | Incorrect Excel sheet name                       | Check the sheet name or use `sheet_name=0`.                           |
| `JSONDecodeError`                 | Invalid JSON format                              | Verify that the JSON file is properly formatted.                      |
| `EmptyDataError`                  | File is empty                                    | Add data to the file or choose a valid file.                          |
| `OSError`                         | Invalid file path or unsupported file            | Check the path and file extension.                                    |
| `ImportError`                     | Required library is missing                      | Install the required package using `pip install package_name`.        |

```python
import pandas as pd
# LOC
data = {
    "Name": ["Aman", "Riya", "Karan", "Neha"],
    "Age": [20, 21, 19, 22],
    "Marks": [85, 90, 78, 95]
}

df = pd.DataFrame(data)

print(df.loc[1])                 # Row with label 1
print(df.loc[0:2])               # Rows 0 to 2 (inclusive)
print(df.loc[:, "Name"])         # Name column
print(df.loc[1:3, ["Name", "Marks"]])

# ILOC

print(df.iloc[1])             # Second row
print(df.iloc[0:3])           # Rows 0,1,2
print(df.iloc[:, 0])          # First column
print(df.iloc[1:3, 0:2])      # Rows 1-2, Columns 0-1
```

```python
import pandas as pd

# =====================================================
# Create Sample DataFrame
# =====================================================

df = pd.DataFrame({
    "Name": ["Aman", "Riya", "Karan", "Neha", "Arjun"],
    "Age": [20, 21, 19, 22, 20],
    "Marks": [85, 90, 78, 95, 88]
})

print("Original DataFrame")
print(df)


# =====================================================
# 1. isin() Examples
# =====================================================

# Example 1: Filter by Name
print("\nNames: Aman and Neha")
print(df[df["Name"].isin(["Aman", "Neha"])])



# Example 2: Filter by Age
print("\nAge: 19 and 22")
print(df[df["Age"].isin([19, 22])])


# Example 3: Filter by Marks
print("\nMarks: 78 and 90")
print(df[df["Marks"].isin([78, 90])])


# =====================================================
# 2. between() Examples
# =====================================================

# Example 1: Marks between 80 and 90
print("\nMarks between 80 and 90")
print(df[df["Marks"].between(80, 90)])


# Example 2: Age between 20 and 22
print("\nAge between 20 and 22")
print(df[df["Age"].between(20, 22)])


# Example 3: Marks between 70 and 80
print("\nMarks between 70 and 80")
print(df[df["Marks"].between(70, 80)])

# =====================================================
# 3. query() Examples
# =====================================================

# Example 1: Age greater than 20
print("\nAge > 20")
print(df.query("Age > 20"))

# Example 2: AND Condition
print("\nAge > 20 AND Marks > 85")
print(df.query("Age > 20 and Marks > 85"))

# Example 3: OR Condition
print("\nAge < 20 OR Marks > 90")
print(df.query("Age < 20 or Marks > 90"))



# Example 4: Equal To
print("\nName == 'Aman'")
print(df.query("Name == 'Aman'"))



# Example 5: Not Equal To
print("\nName != 'Aman'")
print(df.query("Name != 'Aman'"))
```

```python
# Drop rows containing NaN values
print(df.dropna())

# Drop columns containing NaN values
print(df.dropna(axis=1))

# Drop rows only if all values are NaN
print(df.dropna(how="all"))

# Drop rows only if any value is NaN (default)
print(df.dropna(how="any"))


# fillna() Examples

# Fill all NaN values with 0
print(df.fillna(0))

# Fill NaN in a specific column with mean
print(df["Age"].fillna(df["Age"].mean()))

# Fill NaN with a specific value
print(df.fillna("Unknown"))

# Forward Fill (copy previous value)
print(df.ffill())

# Backward Fill (copy next value)
print(df.bfill())
```

```python
# replace() Examples


# Replace a single value
print(df.replace(85, 100))

# Replace multiple values
print(df.replace([85, 90], [100, 95]))

# Replace value in one column
print(df["Name"].replace("Aman", "Rahul"))

# Replace using dictionary
print(df.replace({"Marks": {78: 80, 95: 99}}))



# rename() Examples

# Rename one column
print(df.rename(columns={"Marks": "Score"}))

# Rename multiple columns
print(df.rename(columns={"Name": "Student_Name", "Age": "Student_Age"}))

# Rename row index
print(df.rename(index={0: "A", 1: "B"}))

# Rename permanently
# df.rename(columns={"Marks": "Score"}, inplace=True)
print(df)


# Duplicate Rows Examples
# Check duplicate rows
print(df.duplicated())

# Show duplicate rows
print(df[df.duplicated()])

# Count duplicate rows
print(df.duplicated().sum())

# Remove duplicate rows
print(df.drop_duplicates())

# Remove duplicates based on one column
print(df.drop_duplicates(subset=["Name"]))

# Keep last occurrence
print(df.drop_duplicates(keep="last"))

# Remove duplicates permanently
df.drop_duplicates(inplace=True)
print(df)
```

```python
# =====================================================
# String Cleaning (.str)
# =====================================================

# Convert to lowercase
print(df["Name"].str.lower())

# Convert to uppercase
print(df["Name"].str.upper())

# Capitalize first letter
print(df["Name"].str.capitalize())

# Remove leading/trailing spaces
print(df["Name"].str.strip())

# Remove spaces from left
print(df["Name"].str.lstrip())

# Remove spaces from right
print(df["Name"].str.rstrip())

# Replace text
print(df["Name"].str.replace("Aman", "Rahul"))

# Check if string contains a value
print(df[df["Name"].str.contains("a", case=False)])

# Check starting letters
print(df[df["Name"].str.startswith("A")])

# Check ending letters
print(df[df["Name"].str.endswith("a")])

# Length of each string
print(df["Name"].str.len())

# Split strings
print(df["Name"].str.split(" "))


# =====================================================
# astype() - Data Type Conversion
# =====================================================

# Convert Age to float
print(df["Age"].astype(float))

# Convert Marks to string
print(df["Marks"].astype(str))

# Convert entire DataFrame column
df["Age"] = df["Age"].astype(int)
print(df)

# Convert multiple columns
df = df.astype({
    "Age": "float",
    "Marks": "int"
})
print(df)1
```

```python
# =====================================================
# Sorting
# =====================================================

# Sort by one column (Ascending)
print(df.sort_values(by="Marks"))

# Sort by one column (Descending)
print(df.sort_values(by="Marks", ascending=False))

# Sort by multiple columns
print(df.sort_values(by=["Age", "Marks"]))


# =====================================================
# Adding
# =====================================================

# Add a new column
df["Result"] = "Pass"
print(df)

# Add a column using existing column
df["Bonus"] = df["Marks"] + 5
print(df)

# Insert a column at a specific position
df.insert(1, "City", ["Delhi", "Mumbai", "Pune", "Chandigarh", "Jaipur"])
print(df)


# =====================================================
# Updating
# =====================================================

# Update entire column
df["Marks"] = df["Marks"] + 10
print(df)

# Update a specific value
df.loc[0, "Marks"] = 100
print(df)

# Update using condition
df.loc[df["Marks"] < 40, "Result"] = "Fail"
print(df)

# Update multiple columns
df.loc[1, ["Age", "Marks"]] = [22, 95]
print(df)


# =====================================================
# Deleting
# =====================================================

# Delete a column
print(df.drop("Bonus", axis=1))

# Delete multiple columns
print(df.drop(["Bonus", "Result"], axis=1))

# Delete a row
print(df.drop(0))

# Delete multiple rows
print(df.drop([0, 2]))

# Delete duplicate rows
print(df.drop_duplicates())
```

```python
# =====================================================
# map() Examples
# =====================================================

# Map using dictionary
print(df["Score"].map({95:"A", 85:"B", 92:"A", 78:"C", 88:"B", 96:"A", 82:"B", 75:"C"}))

# Map using lambda
df["Result"] = df["Score"].map(lambda x: "A" if x >= 90 else "B")
print(df)

# Map names to cities
print(df["Name"].map({
    "Alice": "Delhi",
    "Bob": "Mumbai",
    "Charlie": "Pune",
    "David": "Chandigarh",
    "Emma": "Jaipur",
    "Frank": "Lucknow",
    "Grace": "Bhopal",
    "Henry": "Patna"
}))


# =====================================================
# apply() Examples
# =====================================================

# Add 5 to Score
print(df["Score"].apply(lambda x: x + 5))

# Convert names to uppercase
print(df["Name"].apply(lambda x: x.upper()))

# Pass/Fail
print(df["Score"].apply(lambda x: "Pass" if x >= 40 else "Fail"))

# Maximum value of each column
print(df.apply(max))

# Sum of Age and Score for each row
print(df[["Age", "Score"]].apply(sum, axis=1))
```
