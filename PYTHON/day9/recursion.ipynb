```python
compress_string="aaabbbccccdda"
def compress(s):
    if len(s) == 0:
        return ""
    count = 1
    while len(s) > 1 and s[0] == s[1]:
        count += 1
        s = s[1:]
    return s[0] + str(count) + compress(s[1:])

print(compress(compress_string))
```

```python
n=[10,35,6,7,39,2,4]
def nested_maximmum(x):
    if(x==0):
        return 0
    return max(n[x-1],nested_maximmum(x-1))
print(nested_maximmum(len(n)))
```
