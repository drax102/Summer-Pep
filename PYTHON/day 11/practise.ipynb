```python
# convert a snake case into camel case
a="loud_laliture"
b=a.split("_")
c=b[0]+"".join(word.capitalize() for word in b[1:])
print(c)
```

```python
# convert customer names into title case
customer_names = ["john doe", "jane smith", "bob Johnson"]
title_case_names = [name.title() for name in customer_names]
print(title_case_names)
```
